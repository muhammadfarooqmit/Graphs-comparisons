```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("28 Dec 2025 comprassion Phase I.xlsx")
```

```python
df
```

```python
# to start working on row 2 indexing at 1 
df = df.iloc[2:]
df.columns = ["Metrics", "Baseline", "Chunked Fixed", "Decoupled", "Adaptive Chunked", "HCDS"]
df.set_index("Metrics", inplace=True)
```

```python
df = df.apply(pd.to_numeric)
```

```python
df 
```

# 1. BaseLine vs chunked fixed

```python
df[["Baseline", "Chunked Fixed"]].plot(kind="bar")
plt.title("Baseline vs Chunked Fixed")
plt.xlabel("Metrics")
plt.ylabel("Value")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

# 2. Baseline vs chunked fixed vs adaptive chunked

```python
df[["Baseline", "Chunked Fixed", "Adaptive Chunked"]].plot(kind="bar")
plt.title("Baseline vs Chunked Fixed vs Adaptive Chunked")
plt.xlabel("Metrics")
plt.ylabel("Value")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

# 3. Baseline vs decoupled 

```python
df[["Baseline", "Decoupled"]].plot(kind="bar")
plt.title("Baseline vs Decoupled")
plt.xlabel("Metrics")
plt.ylabel("Value")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

# 4. HCDS vs All previous

```python
df[["Baseline", "Chunked Fixed", "Decoupled", "Adaptive Chunked", "HCDS"]].plot(kind="bar")
plt.title("HCDS vs All Previous Techniques")
plt.xlabel("Metrics")
plt.ylabel("Value")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
